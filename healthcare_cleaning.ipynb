```python
import pandas as pd
import numpy as np
#     STEP 2 
# Load Data 
df = pd.read_csv('healthcare_patient_analytics_seaborn.csv')
df.head(5)

# Check the no of rows and columns in dataset
print(df.shape)

# Display summary information about columns and data types
df.info()

# Check he total number of missing (null) value in each column
df.isnull().sum()

# Generate descriptive statistics for numerical columns
df.describe()

# Check for duplicate rows in the dataset
df.duplicated().sum()

# Convert visit_date column from text (string to datetime formate
df['visit_date'] = pd.to_datetime(df['visit_date'])

# Verify the changes by checking the data types again
df.dtypes

# Save the clean dataset to a new csv file without the index column
df.to_csv('healthcare_cleaned_data.csv', index=False)

#        STEP 3
# Import visulization libraries 
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for clean and professional looking graphs
sns.set_theme(style='whitegrid')

# Create a count plot for age groups
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='age_group', palette='viridis')

# Add title and labels 
plt.title('Distribution of patients by Age Group', fontsize=14)
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Number of Patients', fontsize=12)

# Display the plot 
plt.show()

# Create a bar plot to see average treatment cost by department
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='department', y='treatment_cost', errorbar=None, palette='mako')

# Add title and labels
plt.title('Average Treatment Cost by Department', fontsize=14)
plt.xlabel('Department', fontsize=12)
plt.ylabel('Average Cost', fontsize=12)

# Rotate the department name by slightly so they don't overlap
plt.xticks(rotation=15)

# Display the plot
plt.show()

# Create b box plot to analyze hospital stay across different age groups
plt.figure(figsize=(9,6))
sns.boxplot(data=df, x='age_group', y='length_of_stay_days', palette='pastel')

# Add Titles and labels
plt.title('Length of Hospital stay by Age Group', fontsize=14)
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Length of Stay (Days)', fontsize=12)

# Display the plot
plt.show()

# Select only numerical columns for correlation calculation
numerical_df = df[['length_of_stay_days', 'treatment_cost', 'recovery_score', 'readmission_risk']]

# Calculate the correlation matrix
correlation = numerical_df.corr()

# Create a heatmap visualization
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

# Add title
plt.title('Correlation Matrix of Patient Metrics', fontsize=14)

# Display the plot
plt.show()
```
